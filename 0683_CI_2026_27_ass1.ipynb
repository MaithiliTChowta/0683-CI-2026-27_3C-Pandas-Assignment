1. Create DataFrame and Product of First Two Columns

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']

print(df)
```

2. Row-wise and Column-wise Sum

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])

print(df)
print("Row Sum:")
print(df.sum(axis=1))
print("Column Sum:")
print(df.sum(axis=0))
```

3. Fill NaN with Column Mean

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Before:")
print(df)

df = df.fillna(df.mean())

print("After:")
print(df)
```

4. Drop Rows Containing NaN

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Before:")
print(df)

df = df.dropna()

print("After:")
print(df)
```

5. Data Aggregation – Sum and Mean

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

6. Group by Category and Total Sales

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Pen'],
    'Category': ['A', 'B', 'A', 'C', 'B'],
    'Sales': np.random.randint(100, 1000, 5)
})

print(df)

result = df.groupby('Category')['Sales'].sum()

print(result)
```

7. Merging DataFrames

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 70]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

8. MultiIndex DataFrame – Indexing and Slicing

```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2),
    ('B', 1), ('B', 2)
])

df = pd.DataFrame({'Value': [10, 20, 30, 40]},
                  index=index)

print(df)

print("Indexing:")
print(df.loc['A'])

print("Slicing:")
print(df.loc[('A', 1):('B', 1)])
```

9. MultiIndex – Category and SubCategory Sum

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X', 'Y'],
    'Value': np.random.randint(1, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])

print(df)

print("Sum:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

10. Apply Function to Double Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

print("Before:")
print(df)

result = df.apply(lambda x: x * 2)

print("After:")
print(result)
```

11. Lambda Function to Create Sum Column

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda x: x.sum(), axis=1)

print(df)
```
